### 统计函数

```python
# 求和
arr = np.random.randint(1,20,8)
print(arr)
```

```python
# 求和 
print(np.sum(arr))
# 计算平均值
print(np.mean(arr))
# 计算中位数
print(np.median(arr))
# 计算标准差和方差
print(np.var(arr))
print(np.std(arr))
# 最值及其位置
print(np.max(arr),np.argmax(arr))
print(np.min(arr),np.argmin(arr))
# 分位数
print(np.percentile(arr,0))
# 累计的和 积
print(np.cumsum(arr))
print(np.cumprod(arr))
```

## 比较函数

```python
# 是否大于
print(np.greater([1,2,3,4],2))
#是否小于
print(np.less([1,2,3,4],2))
# 是否等于
print(np.equal([1,2,3,4],[2,2,2,2]))
# 逻辑与 逻辑或
print(np.logical_and([1,0],[1,1]))
# 是否有一个元素
print(np.any([0,0,0,1]))
print(np.all([1,1,0]))
# 自定义条件
arr = np.array([1,2,3,4,5])
print(np.where(arr > 3))
```

题目 1:温度数据分析
某城市一周的最高气温('℃)为[28,30,29,31,32,30,29]。
- 计算平均气温、最高气温和最低气温。
- 找出气温超过 30℃ 的天数。

```python
temp = np.array([28,30,29,31,32,30,29])
aver_temp = np.average(temp)
max_temp = np.max(temp)
min_temp = np.min(temp)
days = np.argwhere(temp > 30)
print(f"平均气温{aver_temp:.3f}、最高气温{max_temp}和最低气温{min_temp}，气温超过 30℃ 的天数为{days}")
```

题目 2:学生成绩统计
某班级 5 名学生的数学成绩为[85,90,78,92,88]。

题目 4:随机数据生成
生成一个(3,4)的随机整数数组,范围[0,10)。
- 计算每列的最大值和每行的最小值。
- 将数组中的所有奇数替换为 -1。|

```python
np.random.seed(0)
arr = np.random.randint(0,10,(3,4))
print(arr)
print(f"每行的最大值为{np.max(arr,axis= 1)},最小值为{np.min(arr,axis= 1)}") # 1是列 0是行
print(f"每列的最大值为{np.max(arr,axis= 0)},最小值为{np.max(arr,axis= 0)}")
print(np.where(arr%2 == 0,arr,-1))  # where 本身不改变矩阵
```

题目 5:数组变形
创建一个 1到12 的一维数组,并转换为(3,4)的二维数组。
- 计算每行的和与每列的平均值。
- 将数组展平为一维数组。

```python
arr = np.arange(1,13,dtype=int)
arr = np.split(arr,3)
print(arr)
print(f"每行的和与每列的平均值为{np.sum(arr,axis = 1)},{np.mean(arr,axis=0)}")
print(np.concatenate(arr))
```

题目 20:综合应用
某商店 5 天的销售额(万元)和成本(万元)如下:
销售额:[20,25,22,30,28]
成本:[15,18,16,22,20]
- 计算每天的利润(销售额- 成本)。
-计算利润的平均值和标准差。
- 找出利润最高的天数。


```python
sales = np.array([20,25,22,30,28])
ben = np.array([15,18,16,22,20])
profit = sales - ben
print(f"每天的利润是{sales-ben}")
print(f"利润的平均值为{np.mean(profit)},标准差为{np.std(profit)}")
print(f"利润最高的天数是{np.argmax(profit) + 1}")
```
