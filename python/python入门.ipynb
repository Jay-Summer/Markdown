## 格式化输出

### 含义

#### 三种输出方式

1. 基本格式化输出
    ```python
    print('我的名字是%s,年龄为%d岁' % ("01",18)) #用格式展出，为单引号，后面用括号
    ```
2. format()格式化输出
   ```python
   print("我的名字是:{},年龄为：{}岁".format("01",18))
   ```
3. f表达式格式化输出
   ```python
   name, age = "01", 18
   print(f"我的名字是:{name},年龄为：{age}岁")
   ```

````python
print('我的名字是%s,年龄为%d岁' % ("01",18))
print("我的名字是{},年龄为{}岁".format("01",18))
name, age = "01", 18
print(f"我的名字是:{name},年龄为：{age}岁")
````

````python
# 编写一个99乘法表
i = 1
j = 1
while i <=9 :
    while j <= i:
        print(f"{i}*{j}={i*j}",end='\t')
        j +=1
    i += 1
    j = 1
    print("\n")
````

## 类

- 类属性
- 实例属性
  - `_init_()`来进行初始化
- 类函数 `xxx(self)`
- 类方法 用@classmethod修饰，形参是cls，非动用实例
- 静态方法 用@staticmethod 修饰，没有访问类或者实例

````python
class Person():
    height = 185
    weight = 80
    def __init__(self,name,age):  #双下划线
        self.name = name
        self.age = age
    def sing(self):
        print("aaaaaaa")
    def show(self):
        print("bbbbb")
````

````python
person = Person("01",18)
person.sing()
person.show()
````
